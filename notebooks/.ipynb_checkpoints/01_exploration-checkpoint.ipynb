```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
data = pd.read_csv('../data/raw/HOUSEPRICES.csv')

# Basic exploration
print(data.head())
print(data.info())
print(data.describe())

# Visualizations
plt.figure(figsize=(12,6))
data.hist()
plt.title('House Price Distribution')
plt.show()

# Time series analysis
plt.figure(figsize=(12,6))
data.set_index('date')['price'].plot()
plt.title('House Prices Over Time')
plt.show()
```
